###Loading data from GitHub

```python
# FIND & LOAD CSV FROM GITHUB
import pandas as pd
import matplotlib.pyplot as plt

# 1. RAW URL (Titanic dataset)
csv_url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

# 2. Load CSV
df = pd.read_csv(csv_url)

# 3. Print shape
print("=== DATASET INFO ===")
print(f"File: Titanic.csv")
print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")
print(f"Memory: {df.memory_usage().sum() / 1024:.1f} KB")

# 4. Plot Age column
print("\n=== PLOT ===")
df['Age'].hist(bins=30, color='lightgreen')
plt.title('Passenger Ages on Titanic (GitHub Dataset)')
plt.xlabel('Age')
plt.ylabel('Count')
plt.grid(True, alpha=0.3)
plt.show()

#Show data sample
print("\n=== SAMPLE DATA ===")
print(df[['Name', 'Age', 'Sex', 'Survived']].head(3))
```

```python
#Load a dataset from the cloned repo using a relative path.
import pandas as pd
import os
from glob import glob

# Find CSV files inside the repo (relative paths)
csv_files = glob("**/*.csv", recursive=True)
print("Found CSV files (showing up to 20):")
print("\n".join(csv_files[:20]))

# Pick the first CSV found (or change index)
if not csv_files:
    raise FileNotFoundError("No CSV files found in the cloned repo.")
csv_path = csv_files[0]

df = pd.read_csv(csv_path)
print("Loaded:", csv_path)
print("Shape:", df.shape)
df.head()
```

```python
#Open one notebook from the cloned repo.
from glob import glob

ipynb_files = glob("**/*.ipynb", recursive=True)
print("Found notebooks (showing up to 20):")
print("\n".join(ipynb_files[:20]))

if not ipynb_files:
    raise FileNotFoundError("No .ipynb notebooks found in this repo.")

notebook_to_open = ipynb_files[0]
print("\nOPEN THIS NOTEBOOK in Colab Files pane:")
print(notebook_to_open)
```
